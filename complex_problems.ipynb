```python
from pyspark.sql.functions import *
df_orders = spark.read.table("samples.tpch.orders")
df_suppliers = spark.read.table("samples.tpch.supplier")
df_line_items = spark.read.table("samples.tpch.lineitem").withColumn("revenue", col("l_extendedprice") * (1 - col("l_discount")))
```

```python
df_line_items.groupBy(col("l_suppkey")).agg(sum("revenue").alias("total_revenue")).join(df_suppliers,on = df_suppliers.s_suppkey == df_line_items.l_suppkey, how = "right").groupBy("l_suppkey").agg(sum("total_revenue").alias("tot_rev")).orderBy("tot_rev", Ascending = False).limit(10).display()
# df_suppliers.join(df_line_items,on = df_suppliers.s_suppkey == df_line_items.l_suppkey, how = "left").display()
```
