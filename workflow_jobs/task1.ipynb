```python
schema = dbutils.widgets.get("schema")
table = dbutils.widgets.get("tables")
# schema = 'tpch'
# table = 'orders'
```

```python
from pyspark.sql.functions import *
df = spark.read.table(f"samples.{schema}.{table}")
cust_id = df.groupBy("o_custkey").agg(max(col("o_totalprice")).alias("maxx"), min(col("o_totalprice")).alias("minn")).orderBy("maxx", ascending = False).select("o_custkey").limit(5).collect()
cust_id = [i[0] for i in cust_id]
cust_id
```

```python
dbutils.jobs.taskValues.set("userid",cust_id)
```
